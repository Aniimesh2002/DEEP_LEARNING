Question 1: What is the role of filters and feature maps in Convolutional Neural
Network (CNN)?

>>>

In a **Convolutional Neural Network**, filters and feature maps play an important role in extracting useful patterns from input data such as images. A **filter** (also called a kernel) is a small matrix of weights that slides across the input image during the convolution operation. As the filter moves over different regions of the image, it performs element-wise multiplication and summation with the input pixels. This process helps the network detect specific patterns such as edges, textures, or shapes present in the image.

The result produced after applying a filter to the input image is called a **feature map**. A feature map represents the locations where certain features are detected in the image. Each filter generates its own feature map, and different filters learn to detect different types of features. In deeper layers of the CNN, these feature maps become more abstract and represent more complex patterns, allowing the network to understand high-level structures in the data. Thus, filters are responsible for detecting features, while feature maps store the detected patterns that are used by later layers for classification or prediction.


Question 2: Explain the concepts of padding and stride in CNNs(Convolutional Neural
Network). How do they affect the output dimensions of feature maps?

>>
In a **Convolutional Neural Network**, **padding** and **stride** are important parameters that control how the convolution operation moves over the input and how the size of the output feature map is determined.

**Padding** refers to the process of adding extra pixels (usually zeros) around the border of the input image before applying the convolution filter. The main purpose of padding is to preserve the spatial dimensions of the input and ensure that the edge information is not lost during convolution. Without padding, the size of the feature map becomes smaller after each convolution layer because the filter cannot fully overlap with the border pixels. By adding padding, the network can keep more information from the original input and sometimes maintain the same output size as the input.

**Stride** refers to the number of pixels by which the filter moves across the input image during convolution. If the stride is 1, the filter moves one pixel at a time, producing a larger feature map with more detailed information. If the stride is greater than 1, the filter jumps multiple pixels at each step, which reduces the size of the resulting feature map. A larger stride therefore decreases the output dimension and reduces computational cost, but it may also lose some fine details in the data.

Thus, padding mainly helps **control information loss and preserve dimensions**, while stride controls **how much the filter shifts across the input and how large or small the resulting feature map will be**.


Question 3: Define receptive field in the context of CNNs. Why is it important for deep
architectures?

>>>
In a **Convolutional Neural Network**, the **receptive field** refers to the specific region of the input image that influences the output value of a neuron in a feature map. In other words, it represents the portion of the original input that a particular neuron “sees” when producing its output. In the first convolutional layer, the receptive field is usually small because the neuron only processes a small patch of the input image defined by the filter size. As the data passes through deeper layers of the network, the receptive field gradually increases because each neuron receives information from larger areas of the previous feature maps.

The receptive field is important for deep architectures because it allows the network to capture both **local and global features** of the input data. Early layers with small receptive fields detect simple patterns such as edges or textures, while deeper layers with larger receptive fields can understand more complex structures like shapes or objects. This hierarchical feature learning helps the network build a more comprehensive understanding of the input, which improves performance in tasks such as image recognition and object detection.



Question 4: Discuss how filter size and stride influence the number of parameters in a
CNN.
>>>
In a **Convolutional Neural Network**, the **filter size** and **stride** affect how the convolution operation processes the input and how large the resulting feature maps are, which indirectly influences the model complexity.

The **filter size** directly affects the number of parameters in a convolutional layer. A filter is a small matrix of weights that is applied to the input. The number of parameters in a filter is calculated as the product of the filter’s height, width, and the number of input channels, plus a bias term. Larger filters (such as 5×5 or 7×7) contain more weights than smaller filters (such as 3×3), which increases the number of learnable parameters and computational cost. Smaller filters are often preferred in modern CNN architectures because they reduce parameters while still capturing important features.

The **stride**, on the other hand, determines how far the filter moves across the input image during convolution. While stride changes the **size of the output feature map**, it does **not directly change the number of parameters** in the filter because the same filter weights are reused across the entire input. However, a larger stride reduces the spatial size of the output feature maps, which can reduce the computational cost in later layers.

Therefore, filter size directly affects the **number of parameters and model complexity**, while stride mainly affects the **output dimensions and computational efficiency** of the network.


 Question 5: Compare and contrast different CNN-based architectures like LeNet,
AlexNet, and VGG in terms of depth, filter sizes, and performance.

>>>
Different **Convolutional Neural Network** architectures such as **LeNet**, **AlexNet**, and **VGGNet** differ mainly in their depth, filter sizes, and overall performance.

**LeNet** is one of the earliest CNN architectures and was developed for handwritten digit recognition tasks. It has a relatively shallow structure with only a few convolutional and fully connected layers. The network typically uses small filters such as 5×5 and was designed for simple datasets. Because of its limited depth and computational power, its performance is suitable mainly for smaller and less complex image datasets.

**AlexNet** introduced a much deeper architecture compared to LeNet and played a major role in advancing deep learning for image classification. It contains more convolutional layers and uses larger filters in the early layers, such as 11×11 and 5×5. AlexNet also introduced techniques such as ReLU activation and dropout, which improved training speed and reduced overfitting. As a result, it achieved significantly better performance on large-scale image datasets compared to earlier models.

**VGGNet** further improved CNN performance by increasing the network depth while using smaller filters, typically 3×3. Instead of using large filters, VGG stacks multiple small convolution layers to capture complex patterns. This design allows the network to learn more detailed hierarchical features and improves accuracy in image classification tasks. However, VGG networks require a large number of parameters and high computational resources due to their deep structure.

Overall, LeNet is shallow and designed for simpler tasks, AlexNet introduced deeper architectures and improved training techniques, while VGGNet focuses on deeper networks with smaller filters to achieve higher performance on complex image recognition problems.




















```python
'''
Question 6: Using keras, build and train a simple CNN model on the MNIST dataset
from scratch. Include code for module creation, compilation, training, and evaluation.
'''

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize pixel values
x_train = x_train / 255.0
x_test = x_test / 255.0

# Reshape for CNN (add channel dimension)
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# Create CNN model
model = keras.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.1
)

# Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test)

print("Test Accuracy:", test_acc)
```

```python
'''
Question 7: Load and preprocess the CIFAR-10 dataset using Keras, and create a
CNN model to classify RGB images. Show your preprocessing and architecture.
'''

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()

# Preprocessing
# Normalize pixel values (0-255 → 0-1)
x_train = x_train / 255.0
x_test = x_test / 255.0

# Convert labels to categorical
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)

# CNN Model Architecture
model = keras.Sequential([

    layers.Conv2D(32, (3,3), activation='relu', padding='same', input_shape=(32,32,3)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu', padding='same'),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(128, (3,3), activation='relu', padding='same'),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),

    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),

    layers.Dense(10, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
history = model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2
)

# Evaluate the model
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test Accuracy:", test_accuracy)
```

```python
'''Question 8: Using PyTorch, write a script to define and train a CNN on the MNIST
dataset. Include model definition, data loaders, training loop, and accuracy evaluation.
'''
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms

# Data transformation
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# Load dataset
train_dataset = torchvision.datasets.MNIST(
    root='./data', train=True, download=True, transform=transform)

test_dataset = torchvision.datasets.MNIST(
    root='./data', train=False, download=True, transform=transform)

# Data loaders
train_loader = torch.utils.data.DataLoader(
    train_dataset, batch_size=64, shuffle=True)

test_loader = torch.utils.data.DataLoader(
    test_dataset, batch_size=64, shuffle=False)

# CNN Model definition
class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()

        self.conv1 = nn.Conv2d(1, 32, kernel_size=3)
        self.pool = nn.MaxPool2d(2,2)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3)

        self.fc1 = nn.Linear(64*5*5, 128)
        self.fc2 = nn.Linear(128, 10)

        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.pool(self.relu(self.conv2(x)))

        x = x.view(-1, 64*5*5)
        x = self.relu(self.fc1(x))
        x = self.fc2(x)

        return x

# Initialize model
model = CNN()

# Loss and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training loop
epochs = 5

for epoch in range(epochs):
    running_loss = 0.0

    for images, labels in train_loader:
        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {running_loss/len(train_loader)}")

# Accuracy evaluation
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print("Test Accuracy:", 100 * correct / total)
```

```python
'''Question 9: Given a custom image dataset stored in a local directory, write code using
Keras ImageDataGenerator to preprocess and train a CNN model.'''

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models

# Data preprocessing and augmentation
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(rescale=1./255)

# Load images from directory
train_generator = train_datagen.flow_from_directory(
    'dataset/train',
    target_size=(150,150),
    batch_size=32,
    class_mode='binary'
)

validation_generator = val_datagen.flow_from_directory(
    'dataset/validation',
    target_size=(150,150),
    batch_size=32,
    class_mode='binary'
)

# CNN model architecture
model = models.Sequential([

    layers.Conv2D(32, (3,3), activation='relu', input_shape=(150,150,3)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),

    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),

    layers.Dense(1, activation='sigmoid')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train the model
history = model.fit(
    train_generator,
    epochs=3,
    validation_data=validation_generator
)
```

```python
'''Question 10: You are working on a web application for a medical imaging startup. Your
task is to build and deploy a CNN model that classifies chest X-ray images into “Normal”
and “Pneumonia” categories. Describe your end-to-end approach–from data preparation
and model training to deploying the model as a web app using Streamlit.
'''
'''
ANSWER :
To build a system that classifies chest X-ray images as **Normal** or **Pneumonia**, the workflow begins with data preparation. A dataset of labeled chest X-ray images is collected and organized into folders such as *train*, *validation*, and *test*. Images are resized to a fixed dimension (for example 224×224) and normalized so that pixel values are scaled between 0 and 1. Data augmentation techniques such as rotation, zooming, and horizontal flipping can be applied to increase the diversity of the training data and reduce overfitting. These preprocessing steps can be implemented using tools like **Keras** with image generators.

Next, a **Convolutional Neural Network** is designed for classification. The CNN typically consists of several convolutional layers to extract visual features from the X-ray images, followed by pooling layers to reduce dimensionality. After feature extraction, the feature maps are flattened and passed through fully connected layers for classification. The output layer contains a single neuron with a sigmoid activation function that predicts the probability of pneumonia. The model can be compiled using binary cross-entropy as the loss function and an optimizer such as Adam. During training, the model learns patterns in the images by adjusting its weights over multiple epochs, while validation data is used to monitor performance and detect overfitting. Techniques such as dropout, early stopping, and model checkpoints can be used to improve generalization. After training, the model is evaluated on a test dataset using metrics such as accuracy, precision, recall, and confusion matrix.

Once the model performs satisfactorily, it is saved in a file format such as **.h5** or **SavedModel**. The next step is deployment through a web interface using **Streamlit**. In the Streamlit application, users can upload a chest X-ray image through a file uploader widget. The uploaded image is preprocessed in the same way as the training data (resizing, normalization, and reshaping). The saved CNN model is then loaded and used to make a prediction. The application displays the predicted class (Normal or Pneumonia) along with the probability score. Streamlit provides a simple interface for building interactive dashboards and can be deployed locally or on cloud platforms such as Streamlit Cloud or other hosting services.

This end-to-end workflow includes **data collection and preprocessing, CNN model design and training, evaluation of performance, saving the trained model, and deploying it as an interactive web application** so that doctors or users can upload chest X-ray images and receive automated predictions.
'''
# Training Script (train_model.py)

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models

# Image generators for preprocessing
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(rescale=1./255)

train_data = train_datagen.flow_from_directory(
    "chest_xray/train",
    target_size=(224,224),
    batch_size=32,
    class_mode="binary"
)

val_data = val_datagen.flow_from_directory(
    "chest_xray/val",
    target_size=(224,224),
    batch_size=32,
    class_mode="binary"
)

# CNN Model
model = models.Sequential([

    layers.Conv2D(32,(3,3),activation="relu",input_shape=(224,224,3)),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(64,(3,3),activation="relu"),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(128,(3,3),activation="relu"),
    layers.MaxPooling2D(2,2),

    layers.Flatten(),

    layers.Dense(128,activation="relu"),
    layers.Dropout(0.5),

    layers.Dense(1,activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Train model
history = model.fit(
    train_data,
    epochs=10,
    validation_data=val_data
)

# Save trained model
model.save("pneumonia_model.h5")




# app.py

import streamlit as st
import tensorflow as tf
import numpy as np
from PIL import Image

# Load trained model
model = tf.keras.models.load_model("pneumonia_model.h5")

st.title("Chest X-ray Pneumonia Detection")

uploaded_file = st.file_uploader("Upload Chest X-ray Image", type=["jpg","png","jpeg"])

if uploaded_file is not None:

    image = Image.open(uploaded_file)
    st.image(image, caption="Uploaded Image", use_column_width=True)

    # Preprocess image
    img = image.resize((224,224))
    img = np.array(img)/255.0
    img = img.reshape(1,224,224,3)

    prediction = model.predict(img)

    if prediction[0][0] > 0.5:
        st.error("Prediction: Pneumonia Detected")
    else:
        st.success("Prediction: Normal")
```
